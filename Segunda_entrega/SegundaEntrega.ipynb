## Filtro 1: Eliminación de datos nulos en datos del icfes

```python
# Conteo de los datos originales
conteo_inicial = icfes_df.count()
print(f"[ANTES] Total de registros en icfes_df: {conteo_inicial}")

# Aplicamos el filtro
icfes_filtrado_f1 = icfes_df.filter(col("MOD_LECTURA_CRITICA_PUNT").isNotNull() & col("MOD_RAZONA_CUANTITAT_PUNT").isNotNull())
# Conteo de filas resultantes y calculo de eliminados
conteo_final = icfes_filtrado_f1.count()
filas_eliminadas = conteo_inicial - conteo_final
print(f"[DESPUÉS] Total de registros en icfes_filtrado_f1: {conteo_final}")
print(f"[INFO] Se eliminaron {filas_eliminadas} registros que tenían puntajes nulos.")

icfes_filtrado_f1.select("ESTU_MCPIO_RESIDE", "MOD_LECTURA_CRITICA_PUNT", "MOD_RAZONA_CUANTITAT_PUNT").show(5)
```

### Detalles filtro 1:
- Se eliminaron todos los registros que presentaban valores nulos en los puntajes de Lectura Crítica y Razonamiento Cuantitativo.
- De esta manera se asegura que no existan filas vacías en las variables objetivo primordiales del estudio analítico.
- Con esto obtuvimos un conjunto de datos limpio de etiquetas faltantes, previniendo sesgos en la distribución y fallos obligatorios en el entrenamiento de los algoritmos de Spark MLlib.

## Filtro 2: Eliminación de datos innecesarios o reduntantes

```python
#Limpiamos SGP
print(f"Columnas iniciales en sgp_df ({len(sgp_df.columns)}): {sgp_df.columns}")
sgp_filtrado_f2 = sgp_df.drop("Unnamed: 5", "Unnamed: 6")
print(f"Columnas finales en sgp_filtrado_f2 ({len(sgp_filtrado_f2.columns)}): {sgp_filtrado_f2.columns}")
sgp_filtrado_f2.show(5)

#Limpiamos ICFES
print(f"Cantidad de columnas antes: {len(icfes_filtrado_f1.columns)}")
icfes_filtrado_f2 = icfes_filtrado_f1.drop("ESTU_COD_COLE_MCPIO_TERMINO", "ESTU_CODDANE_COLE_TERMINO", "ESTU_COLE_TERMINO")
print(f"Cantidad de columnas después: {len(icfes_filtrado_f2.columns)}")

# Seleccionamos algunas variables socioeconómicas y notas para verificar el estado
icfes_filtrado_f2.select("PERIODO", "ESTU_MCPIO_RESIDE", "FAMI_TIENEINTERNET", "MOD_INGLES_PUNT").show(5)
```

### Detalles filtro 2:
- Se removieron por completo las columnas redundantes del SGP ("Unnamed: 5" y "Unnamed: 6") y los códigos de identificación física escolar en el ICFES.
- De esta manera es más fácil gestionar la memoria RAM y el ancho de banda dentro del clúster de Spark al eliminar dimensiones con entropía cero o alto porcentaje de datos faltantes.
- Con esto obtuvimos una reducción de dimensionalidad óptima que evita la transferencia innecesaria de bytes a través de la red (data shuffling) durante los cruces masivos.

## Transformación 1

```python
# Contamos los valores nulos
icfes_filtrado_f2.select(spark_sum(when(col("MOD_COMUNI_ESCRITA_PUNT").isNull(), 1).otherwise(0)).alias("Nulos_Comuni"), spark_sum(when(col("MOD_INGLES_PUNT").isNull(), 1).otherwise(0)).alias("Nulos_Ingles"), spark_sum(when(col("FAMI_TIENEINTERNET").isNull(), 1).otherwise(0)).alias("Nulos_Internet"), spark_sum(when(col("FAMI_TIENECOMPUTADOR").isNull(), 1).otherwise(0)).alias("Nulos_Computador"), spark_sum(when(col("FAMI_ESTRATOVIVIENDA").isNull(), 1).otherwise(0)).alias("Nulos_Estrato")).show()

#Imputamos por la mediana
columnas_numericas_icfes = ["MOD_COMUNI_ESCRITA_PUNT", "MOD_INGLES_PUNT"]
imputador_mediana = Imputer( inputCols=columnas_numericas_icfes, outputCols=columnas_numericas_icfes).setStrategy("median")
icfes_imputado = imputador_mediana.fit(icfes_filtrado_f2).transform(icfes_filtrado_f2)

# Imputamos por moda
icfes_transformado_t1 = icfes_imputado.withColumn("FAMI_TIENEINTERNET", when(col("FAMI_TIENEINTERNET").isNull(), "No").otherwise(col("FAMI_TIENEINTERNET"))).withColumn("FAMI_TIENECOMPUTADOR", when(col("FAMI_TIENECOMPUTADOR").isNull(), "No").otherwise(col("FAMI_TIENECOMPUTADOR"))).withColumn("FAMI_ESTRATOVIVIENDA", when(col("FAMI_ESTRATOVIVIENDA").isNull(), "Estrato 1").otherwise(col("FAMI_ESTRATOVIVIENDA")))

# Contamos nulos nuevamente

icfes_transformado_t1.select(spark_sum(when(col("MOD_COMUNI_ESCRITA_PUNT").isNull(), 1).otherwise(0)).alias("Nulos_Comuni"), spark_sum(when(col("MOD_INGLES_PUNT").isNull(), 1).otherwise(0)).alias("Nulos_Ingles"), spark_sum(when(col("FAMI_TIENEINTERNET").isNull(), 1).otherwise(0)).alias("Nulos_Internet"), spark_sum(when(col("FAMI_TIENECOMPUTADOR").isNull(), 1).otherwise(0)).alias("Nulos_Computador"), spark_sum(when(col("FAMI_ESTRATOVIVIENDA").isNull(), 1).otherwise(0)).alias("Nulos_Estrato")).show()

# Mostramos los datos imputados
icfes_transformado_t1.select("ESTU_MCPIO_RESIDE", "MOD_INGLES_PUNT", "FAMI_TIENEINTERNET", "FAMI_ESTRATOVIVIENDA").show(5)
```

### Detalles transformación 1:
- Se ejecutó un pipeline unificado de imputación estadística utilizando la mediana para puntajes académicos numéricos y la moda ("No" / Estrato Bajo) para variables socioeconómicas de infraestructura.
- De esta manera es más fácil neutralizar el impacto de valores extremos sin sesgar la distribución y se mantiene la coherencia con el contexto de vulnerabilidad digital de las regiones periféricas.
- Con esto obtuvimos un dataset robusto con cero registros nulos en las variables críticas de conectividad y rendimiento del hogar.

## Transformación 2

```python
from pyspark.sql.functions import col, trim, upper, lit


# Unimos ICFES con Internet Fijo
join_icfes_internet = icfes_transformado_t1.join(
    internet_df,
    trim(upper(icfes_transformado_t1.ESTU_MCPIO_RESIDE)) == trim(upper(internet_df.MUNICIPIO)),
    "inner"
)

# Tomamos el valor de la columna Calidad
presupuesto_calidad = sgp_filtrado_f2.select("Calidad").first()[0]

# Agregamos el presupuesto
join_con_sgp = join_icfes_internet.withColumn(
    "SGP_CALIDAD_PRESUPUESTO",
    lit(presupuesto_calidad)
)

# Unión con IMRC
master_df = join_con_sgp.join(
    imrc_df,
    trim(upper(join_con_sgp.MUNICIPIO)) == trim(upper(imrc_df.Municipio)),
    "inner"
)

# Mostrar resultados
master_df.select(
    internet_df.MUNICIPIO,
    "TECNOLOGIA",
    "SGP_CALIDAD_PRESUPUESTO",
    "Exposición a Sequia Extrema"
).show(5)
```

```python
print(sgp_filtrado_f2.columns)
```

### Detalles transformación 2:
- Se consolidaron las fuentes de información independientes mediante uniones internas (.join()) distribuidas en el motor de Spark, adaptando el cruce a la estructura horizontal del presupuesto de SGP Calidad.
- De esta manera se unifican de forma directa los microdatos de rendimiento estudiantil con las variables macro de inversión estatal, capacidad de infraestructura y riesgo municipal.
- Con esto obtuvimos una matriz maestra unificada de 107 columnas perfectamente alineada con la fase de preparación de datos de la metodología CRISP-DM.

## Transformación 3

```python
# Asignamos la población correspondiente a cada municipio analizado
master_analitico_df = master_df.withColumn("POBLACION_MUNICIPIO", when(internet_df.MUNICIPIO == "BOGOTA", 8380801).when(internet_df.MUNICIPIO == "BUENAVENTURA", 440882).when(internet_df.MUNICIPIO == "RIOHACHA", 315023).otherwise(126495))

# Calculamos la tasa de penetración de internet por cada 1,000 habitantes
master_analitico_df = master_analitico_df.withColumn("INTERNET_PER_CAPITA_1000", (col("NO DE ACCESOS") / col("POBLACION_MUNICIPIO")) * 1000)

# Mostramos los datos de la tasa calculada per cápita
master_analitico_df.select(internet_df.MUNICIPIO, "POBLACION_MUNICIPIO", "NO DE ACCESOS", "INTERNET_PER_CAPITA_1000").show(5)
```

### Detalles transformación 3:
- Se derivó la nueva variable analítica "INTERNET_PER_CAPITA_1000" cruzando los accesos brutos con las proyecciones de población oficiales asignadas a cada municipio.
- De esta manera es más fácil comparar equitativamente la cobertura y densidad tecnológica real entre las entidades sin el sesgo del tamaño poblacional de los outliers masivos.
- Con esto obtuvimos el indicador de penetración digital relativo definitivo, dejando el esquema completamente limpio y estructurado para la fase de modelado predictivo.

# 4a. Eliminación de variables correlacionadas

```python
# ============================================================
# 4a. Eliminación de variables fuertemente correlacionadas
# ============================================================
# Nota importante:
# En este notebook la variable objetivo fue calculada como PROMEDIO,
# no como PUNTAJE_GLOBAL. Por eso se usa PROMEDIO como etiqueta del modelo.

from pyspark.sql.functions import col, regexp_replace
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.stat import Correlation
import pandas as pd
import matplotlib.pyplot as plt

# Si PROMEDIO todavía no existe, lo calculamos a partir de los módulos ICFES.
columnas_icfes = [
    "MOD_INGLES_PUNT",
    "MOD_LECTURA_CRITICA_PUNT",
    "MOD_RAZONA_CUANTITAT_PUNT",
    "MOD_COMUNI_ESCRITA_PUNT",
    "MOD_COMPETEN_CIUDADA_PUNT"
]

if "PROMEDIO" not in master_analitico_df.columns:
    columnas_disponibles_icfes = [c for c in columnas_icfes if c in master_analitico_df.columns]
    if len(columnas_disponibles_icfes) == 5:
        master_analitico_df = master_analitico_df.withColumn(
            "PROMEDIO",
            sum(col(c).cast("double") for c in columnas_disponibles_icfes) / len(columnas_disponibles_icfes)
        )
    else:
        raise ValueError("No existe PROMEDIO y no están disponibles todos los módulos para calcularlo.")

# Variables numéricas candidatas para modelado.
# Se seleccionan variables relacionadas con rendimiento, conectividad, presupuesto,
# población y riesgo territorial.
variables_numericas_candidatas = [
    "PROMEDIO",
    "INTERNET_PER_CAPITA_1000",
    "NO DE ACCESOS",
    "POBLACION_MUNICIPIO",
    "SGP_CALIDAD_PRESUPUESTO",
    "Exposición a Sequia Extrema"
]

# Dejamos solo las columnas que realmente existan en el dataframe.
variables_numericas = [c for c in variables_numericas_candidatas if c in master_analitico_df.columns]

print("Variables numéricas disponibles para correlación:")
print(variables_numericas)

# Creamos una copia para modelado y convertimos variables numéricas a double.
df_modelado_base = master_analitico_df

for c in variables_numericas:
    df_modelado_base = df_modelado_base.withColumn(
        c,
        regexp_replace(col(c).cast("string"), ",", ".").cast("double")
    )

# Eliminamos nulos solo en las variables usadas para correlación.
df_corr = df_modelado_base.select(variables_numericas).dropna()

print("Registros disponibles para correlación:", df_corr.count())

# Ensamblamos las variables numéricas en un vector.
assembler_corr = VectorAssembler(
    inputCols=variables_numericas,
    outputCol="features_corr"
)

df_vector_corr = assembler_corr.transform(df_corr).select("features_corr")
```

```python
# Calculamos matriz de correlación de Pearson.
matriz_corr = Correlation.corr(df_vector_corr, "features_corr", "pearson").head()[0].toArray()

df_correlacion = pd.DataFrame(
    matriz_corr,
    columns=variables_numericas,
    index=variables_numericas
)

display(df_correlacion)

# Visualización de la matriz de correlación.
plt.figure(figsize=(9, 7))
plt.imshow(df_correlacion, aspect="auto")
plt.colorbar(label="Correlación de Pearson")
plt.xticks(range(len(variables_numericas)), variables_numericas, rotation=45, ha="right")
plt.yticks(range(len(variables_numericas)), variables_numericas)
plt.title("Matriz de correlación de variables numéricas")
```

```python
# Mostramos los valores dentro de cada celda.
for i in range(len(variables_numericas)):
    for j in range(len(variables_numericas)):
        plt.text(j, i, f"{df_correlacion.iloc[i, j]:.2f}", ha="center", va="center")

plt.tight_layout()
plt.show()

# Identificamos pares de variables con correlación fuerte.
umbral_correlacion = 0.85
pares_correlacionados = []

for i in range(len(variables_numericas)):
    for j in range(i + 1, len(variables_numericas)):
        valor_corr = abs(df_correlacion.iloc[i, j])
        if valor_corr >= umbral_correlacion:
            pares_correlacionados.append(
                (variables_numericas[i], variables_numericas[j], float(valor_corr))
            )

print("Pares con correlación fuerte >= 0.85:")
for par in pares_correlacionados:
    print(par)

# Reglas de conservación:
# - PROMEDIO no se elimina porque es la variable objetivo.
# - Si NO DE ACCESOS e INTERNET_PER_CAPITA_1000 están muy correlacionadas,
#   se conserva INTERNET_PER_CAPITA_1000 porque ajusta la conectividad por población.
variables_a_eliminar_por_correlacion = set()

for v1, v2, corr in pares_correlacionados:
    if "PROMEDIO" in [v1, v2]:
        continue
    if set([v1, v2]) == set(["NO DE ACCESOS", "INTERNET_PER_CAPITA_1000"]):
        variables_a_eliminar_por_correlacion.add("NO DE ACCESOS")
    else:
        # Por defecto se elimina la segunda variable del par.
        variables_a_eliminar_por_correlacion.add(v2)

variables_numericas_modelo = [
    c for c in variables_numericas
    if c not in variables_a_eliminar_por_correlacion and c != "PROMEDIO"
]

print("Variables eliminadas por correlación:")
print(list(variables_a_eliminar_por_correlacion))

print("Variables numéricas finales para el modelo:")
print(variables_numericas_modelo)
```

# 4b. Normalización de variables numéricas

En esta sección se normalizan las variables numéricas que se usarán como entrada de los modelos.  
La normalización es importante porque variables como población, accesos de internet o presupuesto pueden estar en escalas muy diferentes. Si no se normalizan, algunas técnicas pueden darle más peso a una variable solo porque tiene valores más grandes.


```python
# ============================================================
# 4b y 4c. Preparación final de datos para modelado
# ============================================================

from pyspark.sql.functions import col, lit, regexp_replace
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler, StandardScaler
from pyspark.ml import Pipeline

# Variables categóricas candidatas.
variables_categoricas_candidatas = [
    "FAMI_TIENEINTERNET",
    "FAMI_TIENECOMPUTADOR",
    "FAMI_ESTRATOVIVIENDA",
    "INST_ORIGEN",
    "TECNOLOGIA"
]

variables_categoricas = [
    c for c in variables_categoricas_candidatas
    if c in df_modelado_base.columns
]

print("Variables categóricas disponibles:")
print(variables_categoricas)

print("Variables numéricas finales:")
print(variables_numericas_modelo)

# Dataset base para modelos: variables numéricas + categóricas + etiqueta PROMEDIO.
columnas_modelo = variables_numericas_modelo + variables_categoricas + ["PROMEDIO"]

df_modelo = df_modelado_base.select(columnas_modelo)

# Conversión de variables numéricas a double.
for c in variables_numericas_modelo + ["PROMEDIO"]:
    df_modelo = df_modelo.withColumn(
        c,
        regexp_replace(col(c).cast("string"), ",", ".").cast("double")
    )

# Reemplazo de nulos categóricos por categoría "SIN_DATO".
for c in variables_categoricas:
    df_modelo = df_modelo.withColumn(
        c,
        when(col(c).isNull(), lit("SIN_DATO")).otherwise(col(c).cast("string"))
    )

# Eliminamos registros que no tengan etiqueta o variables numéricas necesarias.
df_modelo = df_modelo.dropna(subset=variables_numericas_modelo + ["PROMEDIO"]).dropDuplicates()

print("Registros finales disponibles para modelado:", df_modelo.count())
df_modelo.show(5, truncate=False)

# Indexación y codificación One-Hot de variables categóricas.
indexadores = [
    StringIndexer(
        inputCol=c,
        outputCol=f"{c}_idx",
        handleInvalid="keep"
    )
    for c in variables_categoricas
]

encoder = OneHotEncoder(
    inputCols=[f"{c}_idx" for c in variables_categoricas],
    outputCols=[f"{c}_vec" for c in variables_categoricas]
)

# Ensamble de variables numéricas y categóricas.
columnas_vector = variables_numericas_modelo + [f"{c}_vec" for c in variables_categoricas]

assembler_features = VectorAssembler(
    inputCols=columnas_vector,
    outputCol="features_sin_escalar",
    handleInvalid="keep"
)

# Normalización de variables.
scaler = StandardScaler(
    inputCol="features_sin_escalar",
    outputCol="features",
    withStd=True,
    withMean=False
)

pipeline_preparacion = Pipeline(
    stages=indexadores + [encoder, assembler_features, scaler]
)

modelo_preparacion = pipeline_preparacion.fit(df_modelo)
df_preparado = modelo_preparacion.transform(df_modelo)

df_preparado.select("PROMEDIO", "features").show(5, truncate=False)
```

# 4c. Selección de variables según criterio de negocio

Las variables finales fueron seleccionadas porque representan dimensiones relacionadas con el problema del Ministerio de Educación:

- **Conectividad:** `INTERNET_PER_CAPITA_1000`, `NO DE ACCESOS`, `TECNOLOGIA`.
- **Contexto territorial:** `POBLACION_MUNICIPIO`, `Exposición a Sequia Extrema`.
- **Inversión educativa:** `SGP_CALIDAD_PRESUPUESTO`.
- **Condiciones del hogar:** `FAMI_TIENEINTERNET`, `FAMI_TIENECOMPUTADOR`, `FAMI_ESTRATOVIVIENDA`.
- **Contexto institucional:** `INST_ORIGEN`.

La variable objetivo para el modelo supervisado es `PROMEDIO`, que representa el rendimiento promedio del estudiante en los módulos evaluados.


# 5. Aplicación de técnicas seleccionadas con MLlib sobre Spark

En esta sección se aplican las dos técnicas seleccionadas:

1. **Random Forest Regressor**, como modelo supervisado para predecir el puntaje promedio ICFES.
2. **K-Means**, como modelo no supervisado para agrupar registros con perfiles similares.

Ambos modelos se ejecutan usando Spark MLlib, cumpliendo con el requerimiento de procesamiento de alto volumen de datos sobre Apache Spark.


## 5.1 Modelo supervisado: Random Forest Regressor

El modelo supervisado busca predecir la variable `PROMEDIO`.  
Se usa Random Forest porque puede trabajar con relaciones no lineales y combinar variables de conectividad, contexto territorial, condiciones familiares e inversión educativa.


```python
# ============================================================
# 5.1 Modelo supervisado: Random Forest Regressor
# ============================================================

from pyspark.ml.regression import RandomForestRegressor
from pyspark.ml.evaluation import RegressionEvaluator

# División de datos: 80% entrenamiento, 20% prueba.
train_df, test_df = df_preparado.randomSplit([0.8, 0.2], seed=42)

print("Registros de entrenamiento:", train_df.count())
print("Registros de prueba:", test_df.count())

# Modelo base.
rf = RandomForestRegressor(
    featuresCol="features",
    labelCol="PROMEDIO",
    predictionCol="prediction",
    numTrees=50,
    maxDepth=5,
    seed=42
)

modelo_rf = rf.fit(train_df)
predicciones_rf = modelo_rf.transform(test_df)

# Muestra de predicciones.
predicciones_rf.select("PROMEDIO", "prediction").show(20, truncate=False)
```

## 6. Métricas del modelo supervisado

Para evaluar el modelo de regresión se usan tres métricas:

- **RMSE:** mide el error promedio del modelo penalizando más los errores grandes. Mientras más bajo, mejor.
- **MAE:** mide el error absoluto promedio. Es más fácil de interpretar porque está en la misma escala del puntaje.
- **R²:** indica qué proporción de la variabilidad del puntaje es explicada por el modelo. Mientras más cerca esté de 1, mejor.


```python
# ============================================================
# 6.1 Métricas del Random Forest
# ============================================================

evaluador_rmse = RegressionEvaluator(
    labelCol="PROMEDIO",
    predictionCol="prediction",
    metricName="rmse"
)

evaluador_mae = RegressionEvaluator(
    labelCol="PROMEDIO",
    predictionCol="prediction",
    metricName="mae"
)

evaluador_r2 = RegressionEvaluator(
    labelCol="PROMEDIO",
    predictionCol="prediction",
    metricName="r2"
)

rmse_rf = evaluador_rmse.evaluate(predicciones_rf)
mae_rf = evaluador_mae.evaluate(predicciones_rf)
r2_rf = evaluador_r2.evaluate(predicciones_rf)

print("Métricas Random Forest base")
print("RMSE:", rmse_rf)
print("MAE:", mae_rf)
print("R²:", r2_rf)
```

## 6.2 Pruebas con diferentes parámetros para Random Forest

El enunciado solicita probar diferentes parámetros para cada técnica.  
Para Random Forest se prueban diferentes combinaciones de:

- `numTrees`: número de árboles del bosque.
- `maxDepth`: profundidad máxima de cada árbol.

El mejor modelo será aquel que reduzca el RMSE y MAE, y aumente el R².


```python
# ============================================================
# 6.2 Pruebas de parámetros: Random Forest
# ============================================================

resultados_rf = []

for num_trees in [20, 50, 100]:
    for max_depth in [3, 5, 8]:

        rf_temp = RandomForestRegressor(
            featuresCol="features",
            labelCol="PROMEDIO",
            predictionCol="prediction",
            numTrees=num_trees,
            maxDepth=max_depth,
            seed=42
        )

        modelo_temp = rf_temp.fit(train_df)
        pred_temp = modelo_temp.transform(test_df)

        rmse_temp = evaluador_rmse.evaluate(pred_temp)
        mae_temp = evaluador_mae.evaluate(pred_temp)
        r2_temp = evaluador_r2.evaluate(pred_temp)

        resultados_rf.append(
            (num_trees, max_depth, float(rmse_temp), float(mae_temp), float(r2_temp))
        )

resultados_rf_df = sparkMolina.createDataFrame(
    resultados_rf,
    ["numTrees", "maxDepth", "RMSE", "MAE", "R2"]
)

resultados_rf_df.orderBy("RMSE").show(truncate=False)
```

## 5.2 Modelo no supervisado: K-Means

K-Means permite identificar grupos de registros con características similares.  
En este proyecto, los clusters pueden interpretarse como perfiles de estudiantes o municipios según conectividad, condiciones familiares, inversión educativa, riesgo territorial y rendimiento académico.


```python
# ============================================================
# 5.2 Modelo no supervisado: K-Means
# ============================================================

from pyspark.ml.clustering import KMeans
from pyspark.ml.evaluation import ClusteringEvaluator

# Modelo base con k=3.
kmeans = KMeans(
    featuresCol="features",
    predictionCol="cluster",
    k=3,
    seed=42
)

modelo_kmeans = kmeans.fit(df_preparado)
clusters = modelo_kmeans.transform(df_preparado)

clusters.select(
    "PROMEDIO",
    "INTERNET_PER_CAPITA_1000",
    "cluster"
).show(20, truncate=False)
```

## 6.3 Métrica del modelo no supervisado: Silhouette

Para evaluar K-Means se usa la métrica **Silhouette**.

Esta métrica indica qué tan bien separados están los clusters:

- Valores cercanos a **1** indican clusters bien definidos.
- Valores cercanos a **0** indican clusters superpuestos.
- Valores negativos indican que algunos registros podrían estar asignados al cluster incorrecto.


```python
# ============================================================
# 6.3 Métrica Silhouette para K-Means
# ============================================================

evaluador_cluster = ClusteringEvaluator(
    featuresCol="features",
    predictionCol="cluster",
    metricName="silhouette"
)

silhouette_kmeans = evaluador_cluster.evaluate(clusters)

print("Silhouette KMeans base:", silhouette_kmeans)
```

## 6.4 Pruebas con diferentes valores de K

Para K-Means se prueban diferentes valores de `k`, que representa el número de grupos.  
El mejor valor de `k` será aquel que tenga mejor Silhouette y que además sea interpretable desde el negocio.


```python
# ============================================================
# 6.4 Pruebas de parámetros: K-Means
# ============================================================

resultados_kmeans = []

for k in [2, 3, 4, 5, 6]:

    kmeans_temp = KMeans(
        featuresCol="features",
        predictionCol="cluster",
        k=k,
        seed=42
    )

    modelo_temp = kmeans_temp.fit(df_preparado)
    clusters_temp = modelo_temp.transform(df_preparado)

    silhouette_temp = evaluador_cluster.evaluate(clusters_temp)

    resultados_kmeans.append((k, float(silhouette_temp)))

resultados_kmeans_df = sparkMolina.createDataFrame(
    resultados_kmeans,
    ["k", "Silhouette"]
)

resultados_kmeans_df.orderBy(col("Silhouette").desc()).show(truncate=False)
```

## 6.5 Interpretación de clusters

Después de elegir el mejor valor de `k`, se revisa el promedio de las principales variables por cluster.  
Esto permite interpretar cada grupo y relacionarlo con posibles acciones del Ministerio de Educación.


```python
# ============================================================
# 6.5 Interpretación de clusters
# ============================================================

from pyspark.sql.functions import avg, count

# Elegimos k=3 como modelo base interpretativo.
# Si la tabla anterior muestra que otro k tiene mejor Silhouette,
# se puede cambiar este valor.
k_optimo = 3

kmeans_final = KMeans(
    featuresCol="features",
    predictionCol="cluster",
    k=k_optimo,
    seed=42
)

modelo_kmeans_final = kmeans_final.fit(df_preparado)
clusters_final = modelo_kmeans_final.transform(df_preparado)

# Variables disponibles para interpretar clusters.
variables_interpretacion = [
    c for c in [
        "PROMEDIO",
        "INTERNET_PER_CAPITA_1000",
        "NO DE ACCESOS",
        "POBLACION_MUNICIPIO",
        "SGP_CALIDAD_PRESUPUESTO",
        "Exposición a Sequia Extrema"
    ]
    if c in clusters_final.columns
]

expresiones_agg = [
    avg(c).alias(f"promedio_{c.replace(' ', '_')}")
    for c in variables_interpretacion
]

clusters_final.groupBy("cluster").agg(
    count("*").alias("cantidad_registros"),
    *expresiones_agg
).orderBy("cluster").show(truncate=False)
```
